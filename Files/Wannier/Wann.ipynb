```python
import numpy as np
import matplotlib.pyplot as plt

# Read O_D, O_OD, O_TOT values from sprd_down.txt
O_D = []
O_OD = []
O_TOT = []
with open('sprd_down', 'r') as f:
    for line in f:
        if 'O_D=' in line and 'O_OD=' in line and 'O_TOT=' in line:
            parts = line.split()
            od_val = float(parts[1])
            ood_val = float(parts[3])
            otot_val = float(parts[5])
            O_D.append(od_val)
            O_OD.append(ood_val)
            O_TOT.append(otot_val)
O_D = np.array(O_D)
O_OD = np.array(O_OD)
O_TOT = np.array(O_TOT)
```

```python
xmax = 1000
# Plot O_D
plt.figure()
plt.plot(O_D)
plt.xlabel('Step')
plt.ylabel('Dpendence spread values')
plt.grid(True)
plt.yscale("log")
plt.xlim(0, xmax)
plt.show()

# Plot O_OD
plt.figure()
plt.plot(O_OD, color='orange')
plt.xlabel('Step')
plt.ylabel('Independence spread values')
plt.grid(True)
plt.xlim(0, xmax)
convergence_OD = O_OD[2000] if len(O_OD) > 2000 else O_OD[-1]
plt.axhline(y=convergence_OD, color='red', linestyle='--', label=f'Convergence value')
plt.legend()
plt.show()

# Plot O_TOT
plt.figure()
plt.plot(O_TOT, color='green')
plt.xlabel('Step')
plt.ylabel('Total spread values')
plt.grid(True)
plt.xlim(0, xmax)
convergence_TOT = O_TOT[2000] if len(O_TOT) > 2000 else O_TOT[-1]
plt.axhline(y=convergence_TOT, color='red', linestyle='--', label=f'Convergence value')
plt.legend()
plt.savefig('convergence_plot.pdf')
plt.show()
```
