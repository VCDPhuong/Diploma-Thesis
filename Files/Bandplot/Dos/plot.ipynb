```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
DoS = np.array(np.loadtxt('Ru_pdos.dat.pdos_tot', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoSRu1 = np.array(np.loadtxt('Ru_pdos.dat.pdos_atm#1(Ru1)_wfc#4(d)', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoSRu2 = np.array(np.loadtxt('Ru_pdos.dat.pdos_atm#2(Ru2)_wfc#4(d)', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoSO1 = np.array(np.loadtxt('Ru_pdos.dat.pdos_atm#3(O)_wfc#2(p)', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoSO2 = np.array(np.loadtxt('Ru_pdos.dat.pdos_atm#4(O)_wfc#2(p)', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoSO3 = np.array(np.loadtxt('Ru_pdos.dat.pdos_atm#5(O)_wfc#2(p)', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoSO4 = np.array(np.loadtxt('Ru_pdos.dat.pdos_atm#6(O)_wfc#2(p)', skiprows=1, usecols=(0, 1, 2, 3, 4), unpack=True, dtype=float))
DoS = DoS.T
DoSRu1 = DoSRu1.T
DoSRu2 = DoSRu2.T
DoSO1 = DoSO1.T
DoSO2 = DoSO2.T
DoSO3 = DoSO3.T
DoSO4 = DoSO4.T
```

```python
plt.plot(DoS[:, 0], (DoS[:, 1]), label='total DoS up')
plt.plot(DoS[:, 0], (DoS[:, 2]), label='total DoS down')
plt.axvline(x = 12.2400, label='Fermi level', color='black', linestyle='--')
plt.xlim(3,20)
plt.ylim(0,8)
plt.legend()
```

```python
plt.plot(DoS[:, 0], (DoS[:, 1]), label='total DoS up')
plt.plot(DoSRu1[:, 0], (DoSRu1[:, 1]), label='Ru1-d-up')
plt.plot(DoSRu2[:, 0], (DoSRu2[:, 1]), label='Ru2-d-up')
plt.axvline(x = 12.2400, label='Fermi level', color='black', linestyle='--')
plt.xlim(10,14)
plt.ylim(0,8)
plt.legend()
```

```python
plt.plot(DoS[:, 0], (DoS[:, 1]), label='total DoS up')
DoSO = DoSO1 + DoSO2 + DoSO3 + DoSO4
DoSRu = DoSRu1 + DoSRu2
plt.plot(DoS[:, 0], DoSO[:, 1] , label='total O-p-up')
plt.plot(DoS[:, 0], DoSRu[:, 1] , label='total Ru-d-up')
plt.axvline(x = 12.2400, label='Fermi level', color='black', linestyle='--')
plt.xlim(10,14)
plt.ylim(0,8)
plt.legend()
```

```python
plt.plot(DoS[:, 0], (DoS[:, 1] + DoS[:, 2]), label='total DoS')
DoSO = DoSO1 + DoSO2 + DoSO3 + DoSO4
DoSRu = DoSRu1 + DoSRu2
plt.plot(DoS[:, 0], DoSO[:, 1] +  DoSO[:, 2], label='O-p')
plt.plot(DoS[:, 0], DoSRu[:, 1] + DoSRu[:, 2], label='Ru-d')
plt.axvline(x = 12.2400, label='Fermi level', color='black', linestyle='--')
plt.xlim(10,14)
plt.ylim(0,15)
plt.legend()
```
