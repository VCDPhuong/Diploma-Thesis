```python
import pylab as pl
import numpy as np
from matplotlib.gridspec import GridSpec
```

```python
EF = 12.252300
tick_labels = ['$\Gamma$', 'X', 'R', 'Z', '$\Gamma$', 'M', 'A', 'Z']
tick_locs = [0, 0.691905, 1.694871, 2.386776, 3.389743, 4.368243, 5.371210, 6.349711]
```

```python
fig = pl.figure(figsize=(8, 6))
gs = GridSpec(2, 1, hspace=0.00)

# Load data files
databand1 = np.loadtxt('up/RuO2-bands.dat')
databand2 = np.loadtxt('down/RuO2-bands.dat')
data1 = np.loadtxt('up/RuO2-curv.dat')
data2 = np.loadtxt('down/RuO2-curv.dat')
# Top subplot: bands
axes1 = pl.subplot(gs[0, 0:])
pl.scatter(databand1[:,0], databand1[:,1]-EF, color='red', marker='.', s=1.0)
pl.xlim([0, max(databand1[:,0])])
pl.scatter(databand2[:,0], databand2[:,1]-EF, color='blue', marker='.', s=1.0)

pl.ylim([-5,5])
pl.plot([tick_locs[0], tick_locs[-1]], [0, 0], color='red', linestyle='--', linewidth=1.0)
pl.xticks(tick_locs, tick_labels)
for n in range(1, len(tick_locs)):
   pl.plot([tick_locs[n], tick_locs[n]], [pl.ylim()[0], pl.ylim()[1]], color='gray', linestyle='-', linewidth=0.5)
pl.ylabel('Energy$-$E$_F$ [eV]')
pl.tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False)

# Bottom subplot: curvature y
axes2 = pl.subplot(gs[1, 0:])
pl.plot(data1[:,0], data1[:,1], color='red', linestyle='-' ,label='up')
pl.plot(data2[:,0], data2[:,1], color='blue', linestyle='--',label='down')
pl.xlim([0, max(data1[:,0])])
#pl.ylim([min(data1[:,1])-0.025*(max(data1[:,1])-min(data1[:,1])), max(data1[:,1])+0.025*(max(data1[:,1])-min(data1[:,1]))])
pl.ylim(-10,10)
pl.xticks(tick_locs, tick_labels)
for n in range(1, len(tick_locs)):
   pl.plot([tick_locs[n], tick_locs[n]], [pl.ylim()[0], pl.ylim()[1]], color='gray', linestyle='-', linewidth=0.5)
pl.ylabel(r'$-\Omega_x(\mathbf{k})$  [ bohr$^2$ ]')

outfile = 'RuO2-bands+curv_y.pdf'
pl.savefig(outfile, bbox_inches='tight')
pl.legend(loc='upper right')
pl.show()
```
